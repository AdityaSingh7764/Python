```python
class Point:

  def __init__(self,x,y):
    self.x = x
    self.y = y

  def __str__(self):
    return '<{},{}>'.format(self.x,self.y)     # Structure of the object

  def euclidean_distance(self,other):
    return "Euclidean Distance: {}".format(((self.x-other.x)**2 + (self.y-other.y)**2)**0.5)

  def distance_origin(self):
     return "Distance from origin: {}".format(((self.x)**2 + (self.y)**2)**0.5)
    #return "Distance from origin: {}".format(self.euclidean_distance(Point(0,0)))

class Line:

  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return "{}x + {}y + {} = 0".format(self.A,self.B,self.C)

  def point_on_line(Line,Point):
    if Line.A*Point.x + Line.B*Point.y + Line.C == 0:
      return "Point lies on Line"
    else:
      return "Point doesn't lie on the Line"

  def shortest_distance(Line,Point):
    return ("Shortest Distance: {}".format(abs(Line.A*Point.x + Line.B*Point.y + Line.C)/(Line.A**2 + Line.B**2)**0.5))

Point1 = Point(4,7)
Point2 = Point(1,4)

print(Point1)
print(Point2)
print(Point1.euclidean_distance(Point2))
print(Point1.distance_origin())

Line1 = Line(2,4,-36)
print(Line1)
print(Line1.point_on_line(Point1))
print(Line1.point_on_line(Point2))
print(Line1.shortest_distance(Point1))
```

# **2. Reference Variable**

*  **Reference variables hold the objects.**
*  **We can create objects without reference variable**
*  **An object can have multiple reference variable**
*  **Assigning a new reference variable to an existing object does not create a new object**  

```python
class Person:

  def __init__(self):
    self.name="Aditya"
    self.gener="Male"

Person()    # Object without Reference
P = Person()   # P is not the object it is refrence to the address of object
```

```python
# Multiple Reference

Q = P
print(id(P))
print(id(Q))
```

```python
# Change Attribute Value with the help of 2nd Object

print(P.name)
print(Q.name)

Q.name = 'Riya'
print(Q.name)
print(P.name)
```

# **4. Encapsulation**

```python
# ATM Machine

class Atm:

  # Constructor (Special Function)
  def __init__(self):
    self.pin = ''
    self.__balance = 0   # private variable (__)
    # self.menu()

  def get_balance(self):
    return self.__balance

  def set_balance(self,new_value):
    if type(new_value) == int:
      self.__balance = new_value
    else:
      print("Not a valid datatype")

  def __menu(self):
    user_input = input("""
    How can I help
    1. Create Pin
    2. Change Pin
    3. Check Balance
    4. Withdraw
    5. Exit
    """)

    if user_input == '1':
      # Create Pin
      self.create_pin()
    elif user_input == '2':
      # Change Pin
      self.change_pin()
    elif user_input == '3':
      # Check Balance
      self.check_balance()
    elif user_input == '4':
      # Withdraw
      self.user_withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input("Enter Your Pin: ")
    self.pin = user_pin


    user_balance = int(input("Enter Balance: "))
    self.__balance = user_balance

    print("Pin Created Successfully")
    # self.menu()

  def change_pin(self):
    old_pin = input("Enter old pin: ")
    if old_pin == self.pin:
        # Let him/her change the pin
        new_pin=input("Enter new pin: ")
        self.pin=new_pin
        print("Pin Changed Successfully")
        self.menu()

    else:
      print("Cannot change pin")
      self.menu()

  def check_balance(self):
    user_pin=input("Enter you pin: ")
    if user_pin == self.pin:
      print("Your Balance: ",self.__balance)
    else:
      print("Incorrect Pin")

  def user_withdraw(self):
    user_pin=input("Enter the pin: ")
    if user_pin == self.pin:
      user_amount=int(input("Enter withdraw amount: "))
      if user_amount < self.__balance:
        print("Successfully given: ",user_amount)
        print("Remaining balance: ",self.__balance-user_amount)
        self.__balance=self.__balance-user_amount
      else:
        print("Insufficient balance")
        self.menu()
    else:
      print("Incorrect Pin")
      self.menu()
```

# **5. Collection of Objects**

```python
# list of objects

class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

P1 = Person('Aditya','Male')
P2 = Person('Riya','Female')
P3 = Person('Rohan','Male')

L = [P1,P2,P3]  # Can store objects in List

for i in L:
  print(i.name,i.gender)

D = {'P1':P1,'P2':P2,'P3':P3}
print("\n")
for i in D:
  print(i)
  print(D[i])
  print(D[i].name)
  print(D[i].gender)
```

# **6. Static Keyword**

#### **Difference between Instance and Static Variable**

*   Instance Variable has different value for different objects
*   Static Variable has same value across different objects
*   If the variable is initialized using object name (self) then it is a instance variable
*   If the variable is initialized using class name then it is a static variable

```python
# ATM Machine

class Atm:

  __counter = 1

  # Constructor (Special Function)
  def __init__(self):
    self.pin = ''
    self.__balance = 0   # private variable (__)
    self.cid = Atm.__counter
    Atm.__counter = Atm.__counter + 1
    # self.menu()

  # utility functions
  @staticmethod
  def get_counter():
    return Atm.__counter

  def get_balance(self):
    return self.__balance

  def set_balance(self,new_value):
    if type(new_value) == int:
      self.__balance = new_value
    else:
      print("Not a valid datatype")

  def __menu(self):
    user_input = input("""
    How can I help
    1. Create Pin
    2. Change Pin
    3. Check Balance
    4. Withdraw
    5. Exit
    """)

    if user_input == '1':
      # Create Pin
      self.create_pin()
    elif user_input == '2':
      # Change Pin
      self.change_pin()
    elif user_input == '3':
      # Check Balance
      self.check_balance()
    elif user_input == '4':
      # Withdraw
      self.user_withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input("Enter Your Pin: ")
    self.pin = user_pin


    user_balance = int(input("Enter Balance: "))
    self.__balance = user_balance

    print("Pin Created Successfully")
    # self.menu()

  def change_pin(self):
    old_pin = input("Enter old pin: ")
    if old_pin == self.pin:
        # Let him/her change the pin
        new_pin=input("Enter new pin: ")
        self.pin=new_pin
        print("Pin Changed Successfully")
        self.menu()

    else:
      print("Cannot change pin")
      self.menu()

  def check_balance(self):
    user_pin=input("Enter you pin: ")
    if user_pin == self.pin:
      print("Your Balance: ",self.__balance)
    else:
      print("Incorrect Pin")

  def user_withdraw(self):
    user_pin=input("Enter the pin: ")
    if user_pin == self.pin:
      user_amount=int(input("Enter withdraw amount: "))
      if user_amount < self.__balance:
        print("Successfully given: ",user_amount)
        print("Remaining balance: ",self.__balance-user_amount)
        self.__balance=self.__balance-user_amount
      else:
        print("Insufficient balance")
        self.menu()
    else:
      print("Incorrect Pin")
      self.menu()
```
