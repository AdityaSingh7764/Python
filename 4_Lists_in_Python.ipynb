### **What are Lists ?**

*  List is a data type where you can store multiple items under 1 name. More technically, lists act like dynamic arrays which means you can add more items on the fly.

```python
L=[2,"Aditya",5.2,[12,"Riya",70.2]]
print(L)
```

### **Difference b/w Lists and Arrays**
<br>


*  Fixed Vs Dynamic Size (Array are of fixed size when we declare it while list has dynamic size.)
*  Hetrogeneous (Different types of data can be stored in single list while Array can store data of same datatype.)
*   Speed of Execution (Arrays are generally faster when it comes to accessing individual elements, while lists are faster when it comes to adding or removing elements.)
*   Memory (Arrays are more memory efficient as they contiguously store elements of the same type. Lists, being more flexible, consume more memory.)



### **Characterstics of a List**

*   Ordered

*   Hetrogeneous
*   Can have duplicates

### **Creating a List**

```python
print([]) #Empty List
print([13,16.7,183]) #1D List
print([16,12.5,[12,18.6]]) #2D List
print([12,15.7,[[12,14],[16,18],20]]) #3D List
print([1,3.6,True,5+7j,"Pear"]) #Hetrogenous
print(list("Apple")) #Conversion
```

### **Accessing Items from a List**

```python
# Indexing
L=[1,2,3,4,5,[6,7,8]]
print(L[2]) #positive
print(L[5][1]) #nested list access
print(L[-1]) #negative
```

```python
# Slicing
L=[1,2,3,4,[5,6,7]]
print(L[0:3])
print(L[2:])
print(L[1::2])
print(L[::-1])
```

### **Adding items to a list**

```python
# append  (If multiple elements were given then it will take it as a single element to add in list)
L=[1,2,3,4,5]
L.append(6)
print(L)
L.append(False)
print(L)
L.append([7,8,9])
print(L)
```

```python
# extend  (If multiple elements were given then it will take it as multiple elements to add in list)
L=[1,2,3,4,5]
L.extend([6,7,8])
print(L)
L.extend([True,False,6.8])
print(L)
L.extend("Hello")
print(L)
```

```python
# insert  (To insert one element at a particular index in the list)
L=[1,2,3,5]
L.insert(3,4)
print(L)
```

### **Editing items in a list**

```python
L=[1,2,3,4,5] #Using indexing and slicing
L[1]=8
print(L)
L[1:4]=[10,11,12]
print(L)
```

### **Deleting items from a list**

```python
# del
L=[1,2,3,4,5]
print(L)
del L
print(L)
```

```python
# del indexing
L=[1,2,3,4,5]
print(L)
del L[1]
print(L)
```

```python
# del slicing
L=[1,2,3,4,5]
print(L)
del L[0:2]
print(L)
```

```python
# remove (Removes a element given to remove function)
L=[1,2,3,4,5]
print(L)
L.remove(2)
print(L)
```

```python
# pop (Removes the last element from the list is the function is given empty or you can give index of the element that you want to pop from the list.)
L=[1,2,3,4,5]
print(L)
L.pop()
print(L)
L.pop(2)
print(L)
```

```python
# clear (Removes all the elements from the list but doesn't delete the list or makes the list empty)
L=[1,2,3,4,5]
print(L)
L.clear()
print(L)
```

### **Operations on list**
<br>

*   Arithmetic
*   Membership
*   Loop

```python
# Arithmetic (+,*)

L1=[1,2,3,4,5]
L2=[6,7,8,9,10]
print(L1,L2)

# Concatenation
L3=L1+L2
print(L3)

# Repetition
L4=2*L1
print(L4)
```

```python
# Membership (not in, in)

L1=[1,2,3,4,5]
print(2 in L1)
print(6 in L1)
L2=[1,2,3,4,5,[6,7]]
print(6 in L2)
print([6,7] in L2)
```

```python
# Loops
L1=[1,2,3,4,5]
L2=[6,7,8,9,[10,11]]

for i in L2:
  print(i)
```

### **List Comprehension**

* List Comprehension provides a concise way of creating lists.
* Newlist=[Expression for item in
interable if condition == True]
<br>

Advantages of List Comprehension
<br>

* More time efficient and space efficient than loops
* Require fewer lines of code
* Transforms iterative statement into a formula  

```python
# Add 1 to 10 numbers to a list

L=[]
for i in range(1,11):
  L.append(i)
print(L)
```

```python
L=[i for i in range(1,11)]
print(L)
```

```python
# scalar multiplication on a vector
v=[2,3,4]
s=-3
# [-6,-9,12]

L=[s*i for i in v]
print(L)
```

```python
# add squares
L=[1,2,3,4,5]

L=[i**2 for i in L]
print(L)
```

```python
# print all numbers divisible by 5 in range of 1 to 50

L=[i for i in range(1,50) if i%5==0]
print(L)
```

```python
# find languages which start with letter p
Languages=['Java','Python','PHP','C','Javascript']

Languages=[Language for Language in Languages if Language.startswith('P')]
print(Languages)
```

```python
# nested if with list comprehension
Basket=['Apple','Pineapple','Cherry','Banana']
My_Fruits=['Apple','Kiwi','Grapes','Banana']

# add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'

New_Basket=[fruit for fruit in My_Fruits if fruit in Basket if fruit.startswith('A')]
print(New_Basket)
```

```python
# Print a (3,3) matrix using list comprehension

Matrix=[[i*j for i in range(1,4)] for j in range(1,4)]
print(Matrix)
```

```python
# Cartesian Products

L1=[1,2,3,4,5]
L2=[6,7,8,9]

Product=[i*j for i in L1 for j in L2]
print(Product)
```

### **2 Ways to traverse a list**
<br>

*   Itemwise
*   Indexwise



```python
# Itemwise
L=[1,2,3,4,5]

for i in L:
  print(i)
```

```python
# Indexwise
L=[1,2,3,4,5]

for i in range(0,len(L)):
  print(L[i])
```

### **Zip Function**
<br>

The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# Write a program to add items of 2 lists indexwise

L1=[1,2,3,4]
L2=[-1,-2,-3,-4]

list(zip(L1,L2))
```
