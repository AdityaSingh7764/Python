# **1. Nested Loops**

```python
# Print numbers in pairs

n=int(input("Enter the no: "))
for i in range(1,n+1):
  for j in range(1,n+1):
    print(i,j)
  print('\n')
```

```python
# Example- Star Pattern 1

#  *
#  * *
#  * * *
```

```python
n=int(input("Enter the no: "))
for i in range(1,n+1):
  for j in range(1,i+1):
    print("*",end="")
  print("\n")
```

```python
# Example- Pattern 2

#   1
#   121
#   12321
#   1234321
```

```python
n=int(input("Enter the no: "))
for i in range(1,n+1):
  for j in range(1,i+1):
    print(j,end="")
  for k in range(i-1,0,-1):
    print(k,end="")
  print("\n")
```

# **2. Loop Control Statement**



*   Break
*   Continue
*   Pass

```python
# Break Statement

for i in range(1,20):
  if(i==10):
    print(i)
    break
```

```python
for i in range(1,20):
  if(i==5):
    break
  print(i)
```

```python
# Prime Number

lower=int(input("Enter the lower range: "))
upper=int(input("Enter the upper range: "))

for i in range(lower,upper+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i)
```

```python
# Continue Statement

for i in range(1,10):
  if(i==5):
    continue
  print(i)
```

###   **Accessing Substring from a String**

```python
# Indexing (Starts from 0 to n-1)
St="How are You"
print(St[2])
```

```python
# Negative Indexing (Starts from -1 to n)
St="How are You"
print(St[-2])
```

```python
# Slicing (Printing some part of string)
St="How are you"
print(St[1:7])
print(St[1:])
print(St[:7])
print(St[1:10:2])
print(St[-3:])
print(St[::-1])
print(St[7:1:-1])
print(St[-1:-7:-1])
```
