### **a. Aggregation (Has a relation between two classes)**
* **Gives an object to main class as an input**

```python
class Customer:

  def __init__(self,name,gender,address):
    self.name = name
    self.gender = gender
    self.address = address

  def print_add(self):
    print(self.address.city,self.address.pin,self.address.state)

class Address:

  def __init__(self,city,pin,state):
    self.city = city
    self.pin = pin
    self.state = state

Add1 = Address('Meerut',250001,'Uttar Pradesh')
Cust1 = Customer('Aditya','Male',Add1)       # Using another class inside other class is called Aggregation

Cust1.print_add()
```

```python
# Using Private Variables

class Customer:

  def __init__(self,name,gender,address):
    self.name = name
    self.gender = gender
    self.address = address

  def print_add(self):
    print(self.address.get_city(),self.address.get_pin(),self.address.get_state())

class Address:

  def __init__(self,city,pin,state):
    self.__city = city
    self.__pin = pin
    self.__state = state

  def get_city(self):
    return self.__city

  def get_pin(self):
    return self.__pin

  def get_state(self):
    return self.__state

Add1 = Address('Meerut',250001,'Uttar Pradesh')
Cust1 = Customer('Aditya','Male',Add1)

Cust1.print_add()
```

# **3. Super Keyword**

```python
class Phone:

  def __init__(self,price,brand,camera):
    print("Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Buying Phone")

class Smartphone(Phone):

  def buy(self):
    print("Buying Smartphone")
    # Syntax to call parent method
    super().buy()                    # Super keyword is used to call parent variables and methods

S1 = Smartphone(20000,'Samsung',48)
S1.buy()
```

```python
class Phone:

  def __init__(self,price,brand,camera):
    print("Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Buying Phone")

class Smartphone(Phone):

  def buy(self):
    print("Buying Smartphone")

S1 = Smartphone(20000,'Samsung',48)
# S1.super().buy() # Cannot use outside class
```

```python
# super -> constructor

class Phone:

  def __init__(self,price,brand,camera):
    print("Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

class Smartphone(Phone):

  def __init__(self,price,brand,camera,os,ram):
    super().__init__(price,brand,camera)
    self.os = os
    self.ram = ram
    print("Smartphone Constructor")

S1 = Smartphone(20000,'Samsung',48,'Android',32)

print(S1.os)
print(S1.brand)
```

### **Summary**

* A class can inherit from another class.
* Inheritance improves code reuse.
* Constructor, attributes, methods get inherited to the child class.
* The parent has no access to the child class.
* Private properties of parent are not accessible directly in child class.
* Child class can override the attributes or methods. This is called method overriding.
* Super() is an inbuilt function which is used to invoke the parent class methods and constructor.

```python
class Parent:

  def __init__(self,num):
    self.__num = num

  def get_num(self):
    return self.__num

class Child(Parent):

  def __init__(self,num,val):
    super().__init__(num)
    self.__val = val

  def get_val(self):
    return self.val

C1 = Child(10,20)
print(C1.get_num())
print(C1.get_num())
```

```python
class Parent:

  def __init__(self):
    self.num = 10

class Child(Parent):

  def __init__(self):
    super().__init__()
    self.var = 20

  def show(self):
    print(self.num)
    print(self.var)

C1 = Child()
C1.show()
```

# **4. Types of Inheritance**

```python
# Single Inheritance

class Phone:

  def __init__(self,price,brand,camera):
    print("Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Buying Phone")

class Smartphone(Phone):
  pass

Smartphone(20000,'Samsung','48px').buy()
```

```python
# Multilevel Inheritance

class Product:

  def review(self):
    print("Product Customer Review")

class Phone(Product):

  def __init__(self,price,brand,camera):
    print("Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Buying Phone")

class Smartphone(Phone):
  pass

S1 = Smartphone(20000,'Samsung',38)
S1.buy()
S1.review()
```

```python
# Multiple Inheritance

class Phone:

  def __init__(self,price,brand,camera):
    print("Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Buying Phone")

class Product:

  def review(self):
    print("Customer Review")

class Smartphone(Phone,Product):
  pass

S1 = Smartphone(20000,'Samsung','38px')
S1.buy()
S1.review()
```

```python
class A:
  def x1(self):
    return 20

class B(A):
  def x1(self):
    return 30

  def x2(self):
    return 40

class C(B):
  def x2(self):
    return 20

Obj1 = A()
Obj2 = B()
Obj3 = C()
print(Obj1.x1()  + Obj3.x1() + Obj3.x2())
```

```python
class A:

  def x1(self):
    return 20

class B(A):

  def x1(self):
    val = super().x1()+30
    return val

class C(B):

  def x1(self):
    val = self.x1()+20       # Method is calling itself (Infinite Loop)
    return val

Obj1 = C()
print(Obj1.x1())
```

# **5. Polymorphism**
* Method Overriding
* Method Overloading
* Operator Overloading

```python
# Method Overloading

class Shape:

  def area(self,radius):
    return 3.14*radius*radius

  def area(self,l,b):               # Method Overloading is a concept where there are two methods of same name in single class
                                    # ,but there output would be diffrent on the basis of the input.
    return l*b

S1 = Shape()

# S1.area(2)
S1.area(3,4)
```

```python
class Shape:

  def area(self,a,b=0):             # It is generally not available in Python
    if b==0:
      return 3.14*a*a
    else:
      return a*b

S1 = Shape()

print(S1.area(2))
print(S1.area(3,4))
```
