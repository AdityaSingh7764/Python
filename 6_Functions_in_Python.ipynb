# **2. Parameters VS Arguments**

### **Types of Arguments**
<br>

* Default Argument
* Positional Argument
* Keyword Argument

```python
# Default Argument

def power(num1=1,num2=1):
  return num1**num2

power()
```

```python
# Positional Argument

def power(num1=1,num2=1): # The position in which arguments are used by the function.
  return num1**num2

power(2,3)
```

```python
# Keyword Argument

def power(num1=1,num2=1): # You can give arguments by using the variable given in the parameters of a function
  return num1**num2

power(num2=2,num1=3)
```

```python
# *args and **kwargs

# *args and **kwargs are special Python keywords that are used to pass the variable length to a function
```

```python
# *args
# allows us to pass a variable number of non-keyboard arguments to a function
```

```python
def multiply(*args): # Used when there can be n number of arguments to be used in function
  product=1

  for i in args:
    product=product*i

  print(args)
  return product
```

```python
multiply(1,2,3,4,5)
```

```python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.
```

```python
def display(**kwargs):
  for (key,value) in kwargs.items():
    print(key,':',value)
```

```python
display(India='Delhi',Sri_Lanka='Colombo',Nepal='Kathmandu',Pakistan='Islamabad')
```

```python
# Point to remember while using *args and **kwargs
  # Order of the arguments matter (normal->*args->**kwargs)
  # The word "args" and "kwargs" are only a convention, you can use any name of your choice
```

# **4. Variable Scope**

### **Global vs Local Scope**

```python
def gl(b):    # a is a local variable
  print(a)
  print(a+1)
a=4           # b is a global variable
gl(a)
print(a)

#4
#5
#4
```

```python
def fun(b):   # b is a local variable
  a=1         # a value will be 1 in local scope
  a+=1
  print(a)
a=5           # a is a global variable
fun(a)        # a value will be changed to 5 again due to global scope
print(a)

#2
#5
```

# **6. Lambda Function**

**A lambda function is a small anonymous function.**
<br>
**A lambda function can take any number of arguments, but can only have one expression**

#     **lambda a,b: a+b**

```python
# a-> a**2

x=lambda a:a**2
x(2)
```

```python
# a,b->a+b

x=lambda a,b:a+b
x(5,2)
```

### **Difference between lambda vs normal function**
<br>

*   No name   
*   Lambda has no return value (function return a value)
*   Not reusable
*   They are used with Higher Order Function



```python
# check if a string has 'a'

a = lambda s:'a' in s

a('apple')
```

```python
# odd or even

a = lambda x:'even' if x%2==0 else 'odd'

print(a(6))
print(a(7))
```

# **7. Higher Order Function**
<br>


**a function that return a function || a function that take input as a function**

```python
# Example

def square(x):
  return x**2

# HOF
def transform(func, L):
  output=[]
  for i in L:
    output.append(func(i))

  print(output)

L=[1,2,3,4,5]

transform(square,L)
transform(lambda x:x**2,L)
transform(lambda x:x**3,L)
```

### **Map**

```python
# square the items of a list

list(map(lambda x:x**2,[1,2,3,4,5]))
```

```python
# odd/even labelling of list items

L=[1,2,3,4,5]
list(map(lambda x:'even' if x%2==0 else 'odd',L))
```

```python
#fetch names from a list of dict

users =[
    {
        'name':'Aditya',
        'age':21,
        'gender':'male'
    },
    {
        'name':'Riya',
        'age':21,
        'gender':'female'
    },
    {
        'name':'Akash',
        'age':20,
        'gender':'male'
    }
]

list(map(lambda users:users['gender'],users))
```

### **Filter**

```python
# numbers greater than 5

L=[3,4,5,6,7]

list(filter(lambda x:x>5,L))
```

```python
# fetch fruits starting with 'm'

fruits=['apple','litchi','mango']

list(filter(lambda s:s.startswith('m'),fruits))
```

### **Reduce**

```python
import functools

#sum of all item

functools.reduce(lambda x,y:x+y,[1,2,3,4,5])
```

```python
# find min

functools.reduce(lambda x,y: x if x<y else y,[12,34,26,7])
```
