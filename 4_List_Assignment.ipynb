# **2. List Comprehension**

**Q8.	 Create a list of squares of numbers from 1 to 10 using list comprehension.**

```python
[num**2 for num in range(1,11)]
```

**Q9. 	Generate a list containing only even numbers from 1 to 20 using list comprehension.**

```python
[num for num in range(1,21) if num%2==0]
```

**Q10. 	Given a list of words ["apple", "banana", "cherry"], use list comprehension to create a list containing the lengths of each word.**

```python
l=["apple",'banana','cherry']

[len(word) for word in l]
```

**Q11. 	Convert a list of strings ["1", "2", "3"] into a list of integers using list comprehension.**

```python
type(l2[0])
```

**Q12.	 Use list comprehension to filter out only words with more than 4 letters from the list ["cat", "elephant", "dog", "tiger"].**

```python
l=['cat','elephant','dog','tiger']

[words for words in l if len(words)>4]
```

**Q14. 	Given a list nums = [10, 20, 30, 40], write a program to print each item along with its index using for loop and range().**

```python
l=[10,20,30,40]

for num in range(0,len(l)):
  print(l[num])
```

# **4. Zip Function**

**Q16. 	Given two lists:**
**names = ["Alice", "Bob", "Charlie"]**
**scores = [85, 90, 78]**
**Use the zip() function to combine them into a list of tuples.**


```python
names=["Alice","Bob","Charlie"]
scores=[85,90,78]

list(zip(names,scores))
```

**Q17. Use zip() to iterate through two lists [1, 2, 3] and ['a', 'b', 'c'] and print their corresponding elements.**

```python
l1=[1,2,3]
l2=['a','b','c']

for num,alpha in zip(l1,l2):
  print(num,alpha)
```

**Q18.  Given three lists:**
