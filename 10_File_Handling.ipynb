## **Types of Data for I/O**
* Text- '12345' as sequence of unicode characters.
* Binary- 12345 as sequence of bytes of its binary equivalent
<br><br>
 **There are two file types:**
 * Text Files- All program files are text files.
 * Binary Files- Images, music, video, exe files.

## **How File I/O Works**
* Open a file
* Read or Write data
* Close the file

# **1. Writing in a file**

```python
# Case 1: If the file is not present

f = open('example1.txt','w')   # write
f.write("Aditya Singh")
f.close()             # Closes the file to remove it from memory so that no one can access it illegally
# f.write("Hello")    # As file is closed it will not work
```

```python
f = open('example2.txt','w')
f.write("Hi Aditya")
f.write("\nHow are you ?")
f.close()
```

```python
# Case 2: If the file is already present

f = open('example1.txt','w')           # Will replace older content to new content
f.write("Hello Everyone")
f.close()
```

## **Problem with w-mode**
* Replaces older data to new data

## **Solution for w-mode**

```python
# Append Mode

f = open('example1.txt','a')       # Adds new data with previous stored data
f.write("\nHow are you")
f.close()
```

## **To write multiple lines**

```python
L = ["Hello Everyone","\nHow are you","\nMy name is Aditya"]

f = open('/content/example1.txt','w')
f.writelines(L)  # used writeline function instead of write
f.close()
```

# **2. Reading in a file**

```python
# Using read()

f = open('/content/example1.txt','r')
d = f.read()
print(d)
f.close()
```

```python
# Reading upto N characters
# N=10

f = open('/content/example1.txt','r')
d = f.read(10)                          # Upto N characters
print(d)
f.close()
```

```python
# Reading line by line

f = open('/content/example1.txt','r')
print(f.readline(),end='')                          # reading lines using readline
print(f.readline(),end='')                          # readline automatically adds \n
f.close()
```

```python
# Reading entire file using readline

f = open('/content/example1.txt','r')

while True:

  data = f.readline()

  if data == '':
    break
  else:
    print(data,end='')

f.close()
```

# **3. Context Manager (With)**
* It's a good idea to close a file after usage as it will free up resources
* If we don't close it, garbage collector would close it
* with keyword closes the file as soon as the usage is over

```python
# with

with open('/content/example2.txt','w') as f:
  f.write('Hello World')
```

```python
# read with with

with open('/content/example2.txt','r') as f:
  print(f.read())
```

```python
with open('/content/example2.txt','r') as f:
  print(f.readline())
```

```python
with open('example1.txt','r') as f:
  print(f.read(10),end='')                  # 10 char will be printed
  print(f.read(10),end='')                  # Uses buffer memory to resume from that same location
```

```python
# Benefit of with method to load a big file is that you can load a big file which is bigger then your memory by using chucks loading.

big_L = ['hello world' for i in range(1000)]

with open('big_file.txt','w') as f:
  f.writelines(big_L)
```

```python
with open('/content/big_file.txt','r') as f:

  size_of_chunk = 50

  while len(f.read(size_of_chunk)) > 0:
    print(f.read(size_of_chunk),end=' || ')
    f.read(size_of_chunk)
```

```python
# tell function

with open('/content/example1.txt','r') as f:
  print(f.read(10))   # 0-9 character printed
  print(f.tell())    # 10th character
```

```python
# seek function (read)

with open('/content/example1.txt','r') as f:
  print(f.read(10))
  print(f.tell())
  f.seek(0)                             # pointer goes to 0th position to read
  print(f.read(10))
  print(f.tell())
  f.seek(20)
  print(f.read(10))
  print(f.tell())
```

```python
# seek function (write)

with open('/content/example1.txt','w') as f:
  f.write('Hello    Name is Aditya')
  f.seek(6)
  f.write('My')
```

### **Problem with working in text mode**
* Can't work with binary files like images, videos.
* Not good for other data types like list, tuple, float, int.

```python
# working with other datatypes

with open('sample.txt','w') as f:
  # f.write(5) # will not work as other datatype like tuple, list, int\
    f.write('5')
```

```python
with open('sample.txt','r') as f:
  # print(f.read() + 5)               # as str doesn't support string and int addition
    print(int(f.read())+5)
```

```python
# working with complex data

d = {
      'name' : 'Aditya',
      'age' : 21,
      'gender' : 'Male'
}

with open('/content/sample.txt','w') as f:
  # f.write(d)
    f.write(str(d))                                    # Dict doesn't supported by Text File
```

```python
with open('/content/sample.txt') as f:
    print(f.read())
    print(type(f.read()))
    print(dict(f.read()))       # cannot convert string to dict
```

# **5. Serialization and Deserialization**
* **Serialization** - Process of converting python data types to JSON format
* **Deserialization** - Process of converting JSON to python data types

```python
# Serialization using json module
import json

#List
L=[1,2,3,4]

with open('list.json','w') as f:
  json.dump(L,f)
```

```python
#Dict
d = {
      'name' : 'Aditya',
      'age' : 21,
      'gender' : 'Male'
}

with open('dict.json','w') as f:
  json.dump(d,f,indent=4)
```

```python
# Deserialization
import json

with open('dict.json','r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

```python
# Serialize and Deserialize Tuple
import json

t = (1,2,3,4,5)

with open('demo.json','w') as f:
  json.dump(t,f)
```

```python
# Serialize and Deserialize Nested Dict
import json

d = {
    'name':'Aditya',
    'marks':[78,83,92,85]
}

with open('demo.json','w') as f:
  json.dump(d,f)
```

# 6. Serializing and Deserializing Custom Objects

```python
class Person:

  def __init__(self,fname,lname,age,gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

    # format to print in -> Aditya Singh, age -> 21, gender -> Male
```

```python
person = Person('Aditya','Singh',21,'Male')
```

```python
# As a string
import json

def show_object(person):
  if isinstance(person,Person):
    return "{} {} age -> {} gender -> {}".format(person.fname,person.lname,person.age,person.gender)

with open('/content/demo.json','w') as f:
  # print(json.dump(person,f))
  json.dump(person,f,default=show_object)
```

```python
# As a dict
import json

def show_object(person):
  if isinstance(person,Person):
    return {'name':person.fname  + ' ' + person.lname, 'age':person.age, 'gender':person.gender}

with open('/content/demo.json','w') as f:
  json.dump(person,f,default=show_object,indent=4)
```

```python
# Deserializing
import json

with open('/content/demo.json','r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

# **7. Pickling**
**Pickling** is the process whereby a Python object hierarchy is converted into a byte stream, and **unpickling** is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

```python
class Person:

  def __init__(self,name,age):
    self.name = name
    self.age = age

  def display_info(self):
    print('Hi my name is ',self.name,' and I am ',self.age,'years old')
```

```python
p = Person('Aditya',21)
```

```python
# pickle dump
import pickle

with open('person.pk1','wb') as f:          # write binary
  pickle.dump(p,f)
```

```python
# pickle load
import pickle

with open('person.pk1','rb') as f:
  d=pickle.load(f)
  print(d.display_info())
```

# **8. Pickle Vs Json**
* Pickle lets the user to store data in binary format. JSON lets the user data in a human-readable text format
