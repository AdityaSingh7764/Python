# **2. Data Type**

```python
# Integer
print(8)
print(type(8))
# 1*10^308
print(1e308) # max range = 1*10^309
```

```python
# Decimal/Float
print(1.6)
print(type(1.6))
print(1.7e307) # max range = 1.7*10^309
```

```python
# Boolean
print(True)
print(type(True))
print(False)
```

```python
# Complex
print(6+7j)
print(type(6+7j))
```

```python
# String
print("Hello World")
print(type("Hello World"))
```

```python
# List
print([1,5.6,'Hello',True])
print(type([1,5.6,'Hello',True]))
```

```python
# Tuple
print((1,5.6,'Hello',True))
print(type((1,5.6,'Hello',True)))
```

```python
# Sets
print({1,56,167})
print(type({1,56,167}))
```

```python
# Dictionary
print({'Capital':'Delhi','Country':'India','Union Territory':'True'})
print(type({'Capital':'Delhi','Country':'India','Union Territory':'True'}))
```

# **6. Type Conversion**

```python
# Implicit Type Conversion
print(5+3.8)
print(type(5),type(3.8))
```

```python
# Explicit Type Conversion
print(5+int(3.8))
print(type(5),type(int(3.8)))
```

# **7. Literals**

```python
a=0b1001 # Binary
print(a)

b=10 # Decimal
print(b)

c=0o302 # Octal
print(c)

d=0x12 # Hexadecimal
print(d)
```

```python
# Float Literals

float_1=6.4
float_2=2.3e3
float_3=1.3e-2
print(float_1,float_2,float_3)
```

```python
# Complex Literal

complex_1=5.6j
print(complex_1,complex_1.real,complex_1.imag)
```
