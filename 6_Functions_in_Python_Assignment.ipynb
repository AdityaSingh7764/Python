# **3. args and kwargs**

**Q7.	Write a function sum_all that accepts any number of positional arguments using *args and returns their sum.**

```python
def sum_all(*nums):
  sum=0

  for num in nums:
    sum=sum+num

  return sum

sum_all(1,2,3,4,5)
```

**Q8.	Write a function print_details that accepts any number of keyword arguments using **kwargs and prints them in the format "Key: Value".**

```python
def print_details(**keyvalue):

  for key,value in keyvalue.items():
    print("{}:{}".format(key,value))

  return

print_details(name='Aditya',age=21,gender='Male')
```

**Q9.	Write a function mix_args_kwargs that accepts positional arguments (args) and keyword arguments (kwargs) and prints both.**

```python
def mix_args_kwargs(*args, **kwargs):
    sum=0;
    for num in args:
        sum=sum+num
    print("Sum: ",sum,"\n")
    for key, value in kwargs.items():
        print("{}:{}".format(key,value))


mix_args_kwargs(1, 2, 3, name="Alice", age=30, city="New York")
```

**15.	Write a function calculator that returns another function (add, subtract, etc.) based on an argument passed to it.**

```python
def calculator(func):
  def add(a,b):
    return a+b
  def sub(a,b):
    return a-b
  def multi(a,b):
    return a*b
  def div(a,b):
    return a/b

  if func=="add":
    return add
  elif func=="sub":
    return sub
  elif func=="multi":
    return multi
  else:
    return div

ans_cal=calculator("add")
print(ans_cal(10,6))

ans_cal=calculator("sub")
print(ans_cal(10,6))

ans_cal=calculator("multi")
print(ans_cal(10,6))

ans_cal=calculator("div")
print(ans_cal(10,5))
```

# **8. Higher-Order Functions**

**Q18.	Write a higher-order function apply_operation that takes a function and a list of numbers as arguments, applies the function to each number, and returns the results as a list.**

```python
L=[1,2,3,4,5]

def apply_operation(func,L):
  sq_list=[]
  for num in L:
    result=func(num)
    sq_list.append(result)
  return sq_list

print(apply_operation(lambda x:x**2,L))
print(apply_operation(lambda x:x**4,L))
print(apply_operation(lambda x:x**6,L))

```
