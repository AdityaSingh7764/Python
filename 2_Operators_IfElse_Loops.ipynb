# **1. Operators**



*   Arithmatic Operators
*   Relational Operators

**Arithmatic Operator**

```python
a=7
b=3

print(a+b) # Addition
print(a-b) # Subtraction
print(a*b) # Multiplication
print(a/b)  # Division
print(a//b) # Integer Division
print(a%b)  # Modulus (Gives remainder of Division)
print(a**b) # Exponential (Power of)
```

**Relational Operators**

```python
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b) # Equals To
print(a!=b) # Not Equal
```

**Logical Operators**

```python
print(1 and 0) # AND
print(1 or 0) # OR
print(not 1) # NOT
```

**Bitwise Operators**

```python
# Operates on Binary Value

print(2 & 3) # Bitwise AND
print(2 | 3) # Bitwise OR
print(2 ^ 3) # Bitwise XOR
print(~2) # Bitwise NOT
print(4>>2) # Bitwise Left Shift
print(4<<2) # Bitwise Right Shift
```

**Assignment Operators**

```python
a=5
print(a)

a+=5
print(a)

a*=2
print(a)
```

**Membership Operators**

```python
# in/not in

print("A" in "Aditya")
print("A" not in "Aditya")

print(1 in [1,2,4,6,7])
print(1 not in [1,2,4,6,7])
```

```python
# Elif Condition

Email=str(input("Enter the mail: "))
Password=str(input("Enter the password: "))

if(Email=="adityasingh@gmail.com" and Password=="Aditya123"):
    print("Login successful")
elif(Email=="adityasingh@gmail.com" and Password!="Aditya123"):
    print("Password is incorrect")
    New_Password=str(input("Enter the new password:"))
    print("Login Again")
else:
    print("Email or Password is incorrect")
```

```python
# Nested Conditional Statements

Email=str(input("Enter the mail: "))
Password=str(input("Enter the password: "))

if(Email=="adityasingh@gmail.com" and Password=="Aditya123"):
    print("Login successful")
elif(Email=="adityasingh@gmail.com" and Password!="Aditya123"):
    print("Password is incorrect")
    Password=str(input("Enter Correct Password:"))
    if(Password=="Aditya123"): # Nested If and Else
      print("Login Successful")
    else:
        print("Login Again")
else:
    print("Email or Password is incorrect")
```

# **4. Loops**

*   While Loop

```python
# While Loop

# Example - Print a table of a given number

Num=int(input("Enter the number: "))
i=1
while(i<=10):
  print(Num,"*",i,"=",Num*i)
  i+=1
```

```python
# While Loop with Else

i=1
while(i<5):
  print(i)
  i+=1
else:
  print("End")
```

```python
# 4-Program - Guessing Game. Generate a random integer between 1 and 100
```

```python
import random
num=random.randint(1,100)

guess=int(input("Enter a random number to guess: "))
count=0
while(guess!=num):
    if(guess<num):
        print("Go Higher")
    elif(guess):
        print("Go Lower")
    guess=int(input(""))
    count+=1
else:
        print("Correct Guess, The Number is: ",num,"and counter: ",count)
```
