# **1. Tuples**
<br>

A tuple in python is similar to a list. The difference between tuple and list is that we cannot change assigned whereas we can change the elements of a list.

*   A tuple is an immutable list.
*   A tuple can not be changed in any way once it is created
<br>

**Characterstics**


*   Ordered
*   Unchangeable
*   Allows duplicate

## **Should be Learned :**
<br>

*   Creating tuple
*   Accessing items
*   Editing items
*   Adding items
*   Deleting items
*   Operations on tuples
*   Tuple functions

### **Creating Tuple**

```python
#Empty
tup=()
print(tup)
#Single item tuple
tup=('apple',)
print(tup)
print(type(tup))
#Homogenous tuple
tup=(1,2,3,4,5)
print(tup)
#Hetrogenous tuple
tup=(1,True,14.6,[1,True,'Hello'])
print(tup)
#2D tuple
tup=(1,2,3,4,(5,6,7))
print(tup)
#Using type conversion
tup=tuple('apple')
print(tup)
```

### **Accessing Items**
<br>

*   Indexing
*   Slicing

```python
#indexing
tup=(1,2,3,4,5,[6,7])
print(tup)
print(tup[3])
print(tup[-1])
print(tup[5][1])
```

```python
#slicing
tup=(1,2,3,4,5)
print(tup)
print(tup[0:4])
print(tup[0:4:2])
print(tup[-2:])
print(tup[-1:-4:-1])
```

### **Editing Tuple**

```python
tup=(1,2,3,4,'apple')
#tup[2]='hello'
# As tuple are immutable you cannot change the items in it.
```

### **Adding Items**

```python
tup=(1,2,3,4,'hello',15.6)
print(tup)
# No function like append, insert or extend as tuple are immutable
```

### **Deleting Items**

```python
tup=(1,2,3,4,'hello',15.6)
print(tup)
#you can only delete whole tuple not its items or portion of it.
del(tup)
#print(tup)
```

### **Operations on Tuples**

```python
#Arithmetic + or *
tup1=(1,2,3,4)
tup2=(5,6,7,8)
print(tup1+tup2)
print(2*tup1)
```

```python
#Membership
tup=(1,2,3,4,5)
print(tup)
print(4 in tup)
print(3 not in tup)
```

```python
#Iteration
tup=(1,2,3,4,5)
print(tup)
for i in tup:
  print(i)
```

### **Tuple Functions**

```python
tup=(4,5,1,2,3)
print(tup)

#len
print(len(tup))

#sum
print(sum(tup))

#min
print(min(tup))

#max
print(max(tup))

#sorted
print(sorted(tup))
print(sorted(tup,reverse=True))
```

```python
#count
tup=(1,2,3,4,5,6,1,2,2)
print(tup)
print(tup.count(2))
```

```python
#index
tup=(1,2,3,4,5,6,1,2,2)
print(tup)
print(tup.index(3))
```

# **Difference between Lists and Tuple**

*   Syntax
*   Mutability
*   Speed (Generally Tuples are faster than List)
*   Memory (Tuples take less memory than list)
*   Built in functionality (List gives more functions than tuple)
*   Error prone (Lists are more error prone)
*   Usability (If the data is required to be changed then we use lists if the data is for only reading purpose then we can use tuples.)



### **Special Syntax**

```python
#Tuple Unpacking
a,b,c=(1,2,3)
print(a,b,c)
```

```python
a=2
b=4
print(a,b)
a,b=b,a
print(a,b)
```

```python
a,b,*restofit=(1,2,3,4)
print(a,b)
print(restofit)
```

```python
#zipping tuples
tup1=(1,2,3,4,5)
tup2=(6,7,8)
tuple(zip(tup1,tup2))
```

# **2. Sets**
<br>

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).
<br>
However, a set itself is mutable. We can add or remove items from it.
<br>
Sets can also be used to perform mathematical set operations like union, ingteraction, symmetric difference, etc
<br><br>
**Characterstics**
<tb>
*   Unordered
*   Mutable
*   No Duplicates
*   Can't contain mutable data types

### **Creating Sets**

```python
#empty
s=set()
print(s)
print(type(s))

#1D
s={1,2,3}
print(s)
print(type(s))

#2D
#s={1,2,3,4,{5,6}} #Can't have mutable datatype inside set as set is a mutable datatype itself.
#print(s)

#Homogenous
s={1,2,3,4,5}
print(s)

#Hetrogenous
s={1,2,3.5,True,5+7j,(6,7,8)}
print(s)
type(s)

#using type conversion
s=set({1,2,3,4})
print(s)
type(s)

# Duplicates are not allowed
# s={1,1,2,3,4,5}
# print(s)

# Sets display items unordered
s1={1,2,3}
s2={2,3,1}
print(s1==s2)
```

### **Accessing Items**

```python
#s={1,2,3,4}
#s[1]
#Accessing items doesn't work in set
```

### **Editing Items**

```python
#s={1,2,3,4}
#s[1]=100
#Editing items doesn't work in set
```

### Adding Items

```python
s={1,2,3,4}

#add
s.add(5) #can add single element in the set
print(s)

#update
s.update([6,7,8]) #can add multiple elements in the set
print(s)
```

### **Deleting Items**

```python
#del
s={1,2,3,4}
del s
#print(s)
#del(s[0]) #Can't do as indexing doesn't work

#discard (Will not throw an error if element is not present in the set)
s={1,2,3,4}
s.discard(3)
print(s)

#remove (Will throw an error if element is not present in set)
s={1,2,3,4}
s.remove(3)
print(s)

#pop (Randomly deletes an item from set)
s={1,2,3,4}
s.pop()
print(s)

#clear (Removes all element from the set and gets empty set)
s={1,2,3,4}
s.clear()
print(s)
```

### **Set Operation**

```python
#Union(|)
s1={1,2,3,4,5}
s2={1,2,3,6,7}
print(s1|s2)

#Intersection(&)
print(s1&s2)

#Difference(-) (Items that are not present in first set)
print(s1-s2)
print(s2-s1)

#Symmetric Difference(^) (Items that are not common from both sets)
print(s1^s2)

#Membership Test
print(1 in s1)
print(6 in s1)

#Loops
for i in s1:
  print(i)
```

### **Set Functions**

```python
s1={1,2,3,4,5}

#len
print(len(s1))

#sum
print(sum(s1))

#min
print(min(s1))

#max
print(max(s1))

#sorted
print(sorted(s1))

print(sorted(s1,reverse=True))
```

```python
s1={1,2,3,4,5}
s2={1,2,3,6,7}

#union
print(s1.union(s2))

#update
s1.update(s2)
print(s1)
print(s2)
```

```python
s1={1,2,3,4,5}
s2={1,2,3,6,7}

#intersection
print(s1.intersection(s2))

#intersection update
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
s1={1,2,3,4,5}
s2={1,2,3,6,7}

#difference
print(s1.difference(s2))

#difference update
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
s1={1,2,3,4,5}
s2={1,2,3,6,7}

#symmetric_difference
print(s1.symmetric_difference(s2))

#symmetric_difference_update
s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
#isdisjoint (No items common between both the sets)
s1={1,2,3,4,5}
s2={1,2,3,6,7}

print(s1.isdisjoint(s2))

s1={1,2,3,4,5}
s2={6,7,8,9}

print(s1.isdisjoint(s2))
```

```python
#issubset
s1={1,2,3}
s2={1,2,3,6,7}

print(s1.issubset(s2))

s1={1,2,3,4,5}
s2={1,2,3,6,7}

print(s1.issubset(s2))
```

```python
#issuperset
s1={1,2,3,4}
s2={1,2}

print(s1.issuperset(s2))

s1={1,2,3,4,5}
s2={1,2,3,6,7}

print(s1.issuperset(s2))
```

```python
#copy
s1={1,2,3,4}
s2={5,6,7,8}
s2=s1.copy()
print(s2)
```

# **3. Dictionary**
<br>
Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data type which hold only a single value as an element.
<br>
In some languages it is known as map or associative arrays.
<br>
dict={'Name':'Aditya','Age':21,'Gender':'Male'}
<br><br>

**Characterstics**
<tb>

*   Mutable
*   Indexing has no meaning
*   Keys can't be duplicated
*   Keys can't be mutable items

### **Create Dictionary**

```python
#empty dictionary
d={}
print(d)
print(type(d))

#1D dictionary
d={'Name':'Aditya','Age':21,'Gender':'Male'}
print(d)
print(type(d))

#2D dictionary
d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

#using dict function
d=dict([('Name','Aditya'),('Age',21)])
print(d)

#duplicate keys
d={'Maruti Suzuki':'WagonR','Maruti Suzuki':'Swift'}
print(d)

#mutable items as keys
d={'Fruit':'Apple',(1,2,3):'Banana'}
print(d)
```

### **Accessing Items**

```python
d={'Name':'Aditya','Age':'21'}

#[]
print(d['Name'])
print(d['Age'])

#get
print(d.get('Name'))
print(d.get('Age'))

d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)
print(d['Subjects']['DSA'])
```

### **Adding Key-Value Pair**

```python
d['Gender']='Male'
print(d)

d['Degree']='B.Tech'
print(d)

d['Subjects']['Os']=86
print(d)
```

### **Remove Key-Value Pair**

```python
d={'Name': 'Aditya', 'Age': '21', 'Gender': 'Male', 'Degree': 'B.Tech'}
print(d)

#pop
d.pop('Degree')
print(d)

#popitem (Deletes last key value pair)
d.popitem()
print(d)

#del
del d['Age']
print(d)

d={'Name': 'Aditya', 'Age': '21', 'Gender': 'Male', 'Degree': 'B.Tech'}

#clear (Makes dictionary empty)
d.clear()
print(d)

d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

del d['Subjects']['DSA']
print(d)
```

### **Editing Key-Value Pair**

```python
d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

d['Sem']='7th'
print(d)

d['Subjects']['DAA']=84
print(d)
```

### **Dictionary Operations**
<br>
<tb>

*   Membership
*   Iteration



```python
d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

#membership
print('Name' in d)

print('Subjects,DSA' in d)
```

```python
d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

#iteration
for i in d:
  print(i,d[i])
```

### **Dictionary Functions**

```python
d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

#len
print(len(d))

#sorted (Sorts all keys)
print(sorted(d))

print(sorted(d,reverse=True))

#min
print(min(d))

#max
print(max(d))
```

```python
d={
    'Name':'Aditya',
    'Age':21,
    'Gender':'Male',
    'Sem':'8th',
    'Subjects':
    {
        'DSA':68,
        'DAA':72
    }
  }
print(d)

#items
print(d.items()) #Show all items in the form of tuples inside list

#keys
print(d.keys()) #print all keys

#values
print(d.values()) #print all values
```

```python
#update
d1={1:5,2:6,3:7}
d2={3:8,4:9,5:10}

d1.update(d2)
print(d1)
```

### **Dictionary Comprehension**

```python
#print first 10 numbers and their squares

{i:i**2 for i in range(1,11)}
```

```python
#using existing dictionary create a new dictionary
distances={'Delhi':1000,'Mumbai':2000,'Bangalore':3000}

{key:value*0.62 for (key,value) in distances.items()}
```

```python
#using zip
days=["Sunday","Monday","Wednesday","Thursday","Friday","Saturday"]
date=[16,17,18,19,20,21,22]

{i:j for (i,j) in zip(days,date)}
```

```python
#if condition
subjects={'DSA':68,'DAA':72,'OS':76}

{i:j for (i,j) in subjects.items() if j>=72}
```

```python
#nested comprehension
#print tables of number from 2 to 4

#{
 #2:{1:2,2:4,3:6,4:8,5:10},
 #3:{1:3,2:6,3:9,4:12,5:15},
 #4:{1:4,2:8,3:12,4:16,5:20}
#}
```

```python
{i:{j:i*j for j in range(1,11)} for i in range(2,5)}
```
