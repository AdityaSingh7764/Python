### **There are two stages where error may occur in a program:**
* **During compilation - Syntax Error**
* **During execution - Exceptions**

## **Types of Syntax Error**

### **1. IndexError**
The Index Error is thrown when trying to access an item at an invalid index

```python
# 1. IndexError

L=[1,2,3]
print(L[5])
```

### **2. ModuleNotFoundError**
The ModuleNotFoundError is thrown when a module could not be found

```python
# 2. ModuleNotFoundError

import mats
```

### **3. KeyError**
The KeyError is thrown when a key is not found

```python
# 3. KeyError

d = {'name':'Aditya'}
print(d['age'])
```

### **4. TypeError**
The TypeError is thrown when an operation or function is applied to an object of an inappropriate type.

```python
# 4. TypeError

print(1 + 'a')
```

### **5. ValueError**
The ValueError is thrown when a function's argument is of an inappropriate type.

```python
# 5. ValueError

int('A')
```

### **6. NameError**
The NameError is thrown when an object could not be found

```python
# 6. NameError

print(Data)
```

```python
# 7. Attribute Error

L = [1,2,3,4]
print(L.upper())
```

# **3. Raise(Throw) Exception**
*  In python, exceptions are raised when errors occur at runtime.
*  Also, we can raise error manually in raise exceptions

```python
raise NameError("Variable is not present in file")
```

```python
raise FileNotFoundError("File is not present")
```

```python
class Atm:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if(amount < 0):
      raise Exception("Amount cannot be negative")
    if(amount > self.balance):
      raise Exception("Insufficient Balance")
    self.balance = self.balance - amount

P1 = Atm(8000)
try:
  P1.withdraw(-9000)
except Exception as e:
  print(e)
else:
  print(P1.balance)
```

```python
# Example - Google Sign In from Other Device uses Exception Handling

class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print("Logout")

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError("Another Device Detected")
    if email == self.email and password == self.password:
      print("Welcome")
    else:
      print('Login Error')

S1 = Google('Aditya','aditya123@gmail.com','1234','Android')

try:
  S1.login('aditya123@gmail.com','1234','Windows')
except SecurityError as e:
  e.logout()
else:
  print(S1.name)
finally:
  print("Database Closed")
```
